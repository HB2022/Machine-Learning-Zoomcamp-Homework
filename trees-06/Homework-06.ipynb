```python
import pandas as pd
import numpy as np

import seaborn as sns
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
data = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
```

```python
df = pd.read_csv(data)
```

```python
df.to_csv('Homework-trees_carfuelefficiency_2026.csv', index=False)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df = df.fillna(0)
```

```python
from sklearn.model_selection import train_test_split
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
y_test = df_test.fuel_efficiency_mpg.values
```

```python
del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']
```

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
train_dicts = df_train.to_dict(orient='records')
```

```python
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(max_depth=1, random_state=1)
dt.fit(X_train, y_train)
```

```python
feature_names = dv.get_feature_names_out()

for name, importance in zip(feature_names, dt.feature_importances_):
    if importance > 0:
        print(f"Feature: {name} (Importance: {importance})")
```

```python
#Question1: model_year
```

```python
val_dicts = df_val.to_dict(orient='records')
X_val = dv.transform(val_dicts)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error

rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_val)
rmse = root_mean_squared_error(y_val, y_pred)

print(f"RMSE: {rmse:.3f}")
```

```python
#Question2: RMSE: 1.834
```

```python
scores = []

for n in [10, 50, 100, 150]:
    rf = RandomForestRegressor(n_estimators=n, random_state=1, n_jobs=-1)
    rf.fit(X_train, y_train)
    
    y_pred = rf.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    
    scores.append((n, round(rmse, 3)))
    print(f"n_estimators={n:3d} | RMSE: {rmse:.3f}")
```

```python
#Question3: n_estimators=100 | RMSE: 1.768
```

```python
max_depths = [10, 15, 20, 25]
n_estimators_list = [10, 50, 100, 150]

best_rmse = float('inf')
best_depth = None

for depth in max_depths:
    for n in n_estimators_list:
        rf = RandomForestRegressor(
            max_depth=depth, 
            n_estimators=n, 
            random_state=1, 
            n_jobs=-1
        )
        rf.fit(X_train, y_train)
        
        y_pred = rf.predict(X_val)
        rmse = root_mean_squared_error(y_val, y_pred)
        
        print(f"max_depth={depth:2d} | n_estimators={n:3d} | RMSE: {rmse:.3f}")
        
        if rmse < best_rmse:
            best_rmse = rmse
            best_depth = depth

print(f"\nBest max_depth: {best_depth} with RMSE: {best_rmse:.3f}")
```

```python
#Question4: Best max_depth: 10 with RMSE: 1.746
```

```python
rf = RandomForestRegressor(
    n_estimators=10, 
    max_depth=20, 
    random_state=1, 
    n_jobs=-1
)
rf.fit(X_train, y_train)

feature_names = dv.get_feature_names_out()
importances = rf.feature_importances_

for name, importance in sorted(zip(feature_names, importances), key=lambda x: x[1], reverse=True):
    print(f"{name}: {importance:.4f}")
```

```python
import xgboost as xgb

features = dv.get_feature_names_out().tolist()
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)

watchlist = [(dtrain, 'train'), (dval, 'val')]

xgb_params = {
    'max_depth': 6,
    'min_child_weight': 1,
    'objective': 'reg:squarederror',
    'nthread': 8,
    'seed': 1,
    'verbosity': 1,
}

xgb_params['eta'] = 0.3
model_03 = xgb.train(xgb_params, dtrain, num_boost_round=100, evals=watchlist, verbose_eval=False)
y_pred_03 = model_03.predict(dval)
rmse_03 = root_mean_squared_error(y_val, y_pred_03)

xgb_params['eta'] = 0.1
model_01 = xgb.train(xgb_params, dtrain, num_boost_round=100, evals=watchlist, verbose_eval=False)
y_pred_01 = model_01.predict(dval)
rmse_01 = root_mean_squared_error(y_val, y_pred_01)

print(f"eta=0.3 | Validation RMSE: {rmse_03:.3f}")
print(f"eta=0.1 | Validation RMSE: {rmse_01:.3f}")
```
